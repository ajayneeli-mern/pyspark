#### join to dataframes

```python
from pyspark.sql import SparkSession

# Initialize SparkSession
spark = SparkSession.builder.appName("DataFrame Join Example").getOrCreate()

# Sample data for DataFrame 1
data1 = [("Alice", 34), ("Bob", 45), ("Cathy", 29)]
columns1 = ["Name", "Age"]

# Sample data for DataFrame 2
data2 = [("Alice", "F"), ("Bob", "M"), ("David", "M")]
columns2 = ["Name", "Gender"]

# Create DataFrame 1
df1 = spark.createDataFrame(data1, columns1)

# Create DataFrame 2
df2 = spark.createDataFrame(data2, columns2)

# Perform a join on the "Name" column
joined_df = df1.join(df2, on="Name", how="inner")

# Show the result
joined_df.show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Initialize Spark session
spark = SparkSession.builder \
    .appName("customer_spend") \
    .getOrCreate()

# Sample data
data = [
    (1, 'A', 100),
    (2, 'A', 150),
    (1, 'B', 200),
    (3, 'B', 300),
    (2, 'A', 120),
    (4, 'C', 180),
    (3, 'C', 250),
    (1, 'C', 220)
]

# Create DataFrame
columns = ['customer_id', 'product', 'spend']
df = spark.createDataFrame(data, columns)

# Group by product and customer and calculate total spend
df2 = df.groupBy('product', 'customer_id').agg(F.sum('spend').alias('total_spend'))

# Create a window specification to find the max spend per product
windowSpec = Window.partitionBy('product').orderBy(F.col('total_spend').desc())

# Rank the customers based on their spend for each product
ranked_df = df2.withColumn('rank', F.rank().over(windowSpec))

# Filter to get the customer with the maximum spend for each product
max_spend_per_product = ranked_df.filter(F.col('rank') == 1).drop('rank')

# Show the result
max_spend_per_product.show()

# Show original dataframe
df.show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.window import Window
from pyspark.sql import functions as f
from pyspark.sql import Column

# Create a Spark session
spark = SparkSession.builder.master("local").appName("Running Total").getOrCreate()

# Create the DataFrame
raw_data = [(1, 10), (1, 20), (1, 30), (1, 40), (2, 20), (2, 40), (2, 60), (2, 80)]
df = spark.createDataFrame(data=raw_data, schema=['id', 'total'])

# Define the window specification
windowSpec = Window.partitionBy('id').orderBy('total')

# Calculate the running total
df1 = df.withColumn("running_total", f.sum('total').over(windowSpec))

# Show the result
df1.show()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, regexp_replace

# Initialize Spark session
spark = SparkSession.builder \
    .appName("Mask Sensitive Data") \
    .getOrCreate()

# Read CSV file into a DataFrame
input_path = "path_to_your_csv_file.csv"
df = spark.read.csv(input_path, header=True, inferSchema=True)

# Mask the email addresses by replacing the part before '@' with '****'
df_masked = df.withColumn("email", regexp_replace(col("email"), r"^[^@]+", "****"))

# Show the masked DataFrame
df_masked.show()

# Optionally write the masked DataFrame to a new CSV file
output_path = "path_to_masked_csv_file.csv"
df_masked.write.csv(output_path, header=True)

# Stop the Spark session
spark.stop()
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import when

# Initialize Spark
spark = SparkSession.builder.appName("Change Value Example").getOrCreate()

# Create a sample DataFrame
data = [("John", 25, "Sales"), ("Doe", 30, "HR"), ("Jane", 22, "IT")]
columns = ["Name", "Age", "Department"]

df = spark.createDataFrame(data, columns)

# Show the initial DataFrame
df.show()

# Use `when` to apply a conditional update
df_updated = df.withColumn("Department", 
                           when(df["Name"] == "John", "Marketing").otherwise(df["Department"]))

# Show the updated DataFrame
df_updated.show()
```
